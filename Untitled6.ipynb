```python
import requests, zipfile, io, json
import pandas as pd

# Ask the U.S. Food and Drug Administration (FDA) open data service where the recall files are
index = requests.get("https://api.fda.gov/download.json").json()
files = index["results"]["device"]["recall"]["partitions"]
print("Number of files:", len(files))

# Download each file, unzip it, and collect every recall record
records = []
for f in files:
    r = requests.get(f["file"])
    z = zipfile.ZipFile(io.BytesIO(r.content))
    data = json.loads(z.read(z.namelist()[0]))
    records.extend(data["results"])

print("Total recalls downloaded:", len(records))
```

```python
def first(value):
    # Some fields are lists, so we keep the first item
    if isinstance(value, list):
        return value[0] if value else None
    return value

rows = []
for rec in records:
    extra = rec.get("openfda") or {}
    rows.append({
        "recall_id": rec.get("product_res_number"),
        "date": rec.get("event_date_initiated"),
        "firm": rec.get("recalling_firm"),
        "product": rec.get("product_description"),
        "reason": rec.get("reason_for_recall"),
        "root_cause": rec.get("root_cause_description"),
        "regulation": first(extra.get("regulation_number")),
        "specialty": first(extra.get("medical_specialty_description")),
        "device_class": first(extra.get("device_class")),
    })

df = pd.DataFrame(rows)
df.head()
```

```python
ivd = df[df["regulation"].astype(str).str[:3].isin(["862", "864", "866"])].copy()
print("All device recalls:", len(df))
print("In Vitro Diagnostic (IVD) recalls:", len(ivd))
print()
print(ivd["specialty"].value_counts())
print()
print(ivd["root_cause"].value_counts().head(15))
```

```python
counts = ivd["root_cause"].value_counts()
print(len(counts), "categories in total")
print(counts.to_string())
```

```python
import matplotlib.pyplot as plt

# Your decisions: which root cause belongs to which group
groups = {
    "Design (original)": [
        "Software design", "Device Design", "Component design/selection",
        "Labeling design", "Package design/selection", "Software in the Use Environment"],
    "Design (change)": [
        "Software change control", "Labeling Change Control",
        "Software Design Change", "Component change control"],
    "Labeling error": [
        "Error in labeling", "Labeling False and Misleading",
        "Labeling mix-ups", "Incorrect or no expiration date"],
    "Regulatory": ["No Marketing Application", "PMA"],
    "Use": ["Use error"],
    "Unknown": [
        "Other", "Under Investigation by firm",
        "Unknown/Undetermined by firm", "Pending"],
}

# Turn the dictionary around: root cause -> group
cause_to_group = {cause: g for g, causes in groups.items() for cause in causes}

# Everything not listed above is manufacturing
ivd["group"] = ivd["root_cause"].map(cause_to_group).fillna("Manufacturing")

# Count and calculate percentages
summary = ivd["group"].value_counts()
known = ivd[ivd["group"] != "Unknown"]

design = ivd["group"].str.startswith("Design")
software = ivd["root_cause"].str.contains("Software") & design

print(summary)
print()
print(f"Design share of all recalls:        {design.mean():.0%}")
print(f"Design share of known causes:       {design[known.index].mean():.0%}")
print(f"Software share of design failures:  {software.sum() / design.sum():.0%}")

# Chart
colors = ["#c0392b" if g.startswith("Design") else "#95a5a6" for g in summary.index]
summary.sort_values().plot(kind="barh", color=colors[::-1], figsize=(8, 4))
plt.title("Why are diagnostic devices recalled? (U.S. FDA data, 7,004 recalls)")
plt.xlabel("Number of recalls")
plt.tight_layout()
plt.show()
```
